```python
import cv2
import random
from ai2thor.controller import Controller
```

```python
floor_index = random.randint(0, 30)
floor_index = 1

controller = Controller(
    agentMode="arm",
    visibilityDistance = 1.00,
    scene = f"FloorPlan{floor_index}",

    # step sizes
    snapToGrid = True,
    gridSize = 0.05,

    # image modalities
    renderInstanceSegmentation = False,
    renderDepthImage = False,
    renderSemanticSegmentation = False,
    renderNormalsImage = False,

    width=600,
    height=420,
    fieldOfView=60,

    # set seed for reproducability
    seed = 90,
)

BASE_STEP = 0.05      # 5 cm
ROT_STEP = 10         # 10 degrees
ARM_STEP = 0.05       # 5 cm

print("""
================ ManipulaTHOR Control ================

Mobile base:
    W = forward
    S = backward
    A = move left
    D = move right

    Q = rotate left
    E = rotate right

Camera:
    R = look up
    F = look down

Arm:
    I = arm forward
    K = arm backward

    J = arm left
    L = arm right

    U = arm up
    O = arm down

Interaction:
    G = pickup object
    H = release object

Information:
    P = print robot position
    ESC = quit

======================================================
""")
```

```python
while True:

    # --------------------------------------------------
    # Show current robot camera
    # --------------------------------------------------

    frame = controller.last_event.cv2img

    cv2.imshow(
        "ManipulaTHOR Robot",
        frame
    )

    key = cv2.waitKey(0) & 0xFF


    # ==================================================
    # Exit
    # ==================================================

    if key == 27:
        break

    # ==================================================
    # Initial State
    # ==================================================

    elif key == ord("0"):
        
        # Teleport the robot
        controller.step(
            action="Teleport",
            position={
                "x": -1.0,
                "y": 0.90,
                "z": -1.55
            },
            rotation={
                "x": 0.0,
                "y": 270.0,
                "z": 0.0
            },
            horizon=30.0
        )

        # Reset arm-base height
        controller.step(
            action="MoveArmBase",
            y=0.5
        )
        
        print("Robot and manipulator reset.")

    # ==================================================
    # MOBILE BASE
    # ==================================================

    elif key == ord("w"):
        
        controller.step(
            action="MoveAgent",
            ahead=BASE_STEP,
            right=0,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )


    elif key == ord("s"):
        
        controller.step(
            action="MoveAgent",
            ahead=-BASE_STEP,
            right=0,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )


    elif key == ord("a"):
        
        controller.step(
            action="MoveAgent",
            ahead=0,
            right=-BASE_STEP,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )


    elif key == ord("d"):
        
        controller.step(
            action="MoveAgent",
            ahead=0,
            right=BASE_STEP,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )


    # ==================================================
    # ROTATION
    # ==================================================

    elif key == ord("q"):

        controller.step(
            action="RotateAgent",
            degrees=-ROT_STEP,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )


    elif key == ord("e"):

        controller.step(
            action="RotateAgent",
            degrees=ROT_STEP,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )


    # ==================================================
    # CAMERA
    # ==================================================

    elif key == ord("r"):

        controller.step(
            action="LookUp"
        )


    elif key == ord("f"):

        controller.step(
            action="LookDown"
        )


    # ==================================================
    # ARM
    #
    # coordinateSpace="wrist" means RELATIVE movement
    # ==================================================

    elif key == ord("i"):

        controller.step(
            action="MoveArm",
            position={
                "x": 0,
                "y": ARM_STEP,
                "z": 0
            },
            coordinateSpace="wrist",
            restrictMovement=True,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )

    elif key == ord("k"):

        controller.step(
            action="MoveArm",
            position={
                "x": 0,
                "y": -ARM_STEP,
                "z": 0
            },
            coordinateSpace="wrist",
            restrictMovement=True,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )
        
    elif key == ord("j"):

        controller.step(
            action="MoveArm",
            position={
                "x": -ARM_STEP,
                "y": 0,
                "z": 0
            },
            coordinateSpace="wrist",
            restrictMovement=True,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )

    elif key == ord("l"):

        controller.step(
            action="MoveArm",
            position={
                "x": ARM_STEP,
                "y": 0,
                "z": 0
            },
            coordinateSpace="wrist",
            restrictMovement=True,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )

    elif key == ord("u"):

        controller.step(
            action="MoveArm",
            position={
                "x": 0,
                "y": 0,
                "z": ARM_STEP
            },
            coordinateSpace="wrist",
            restrictMovement=True,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )

    elif key == ord("o"):

        controller.step(
            action="MoveArm",
            position={
                "x": 0,
                "y": 0,
                "z": -ARM_STEP
            },
            coordinateSpace="wrist",
            restrictMovement=True,
            returnToStart=True,
            speed=1,
            fixedDeltaTime=0.02
        )

    # ==================================================
    # PICKUP
    # ==================================================

    elif key == ord("g"):

        event = controller.step(
            action="PickupObject"
        )

        print(
            "Pickup:",
            event.metadata["lastActionSuccess"]
        )

    # ==================================================
    # RELEASE
    # ==================================================

    elif key == ord("h"):

        event = controller.step(
            action="ReleaseObject"
        )

        print(
            "Release:",
            event.metadata["lastActionSuccess"]
        )

    # ==================================================
    # PRINT ROBOT STATE
    # ==================================================

    elif key == ord("p"):

        event = controller.last_event

        agent = event.metadata["agent"]
        arm = event.metadata["arm"]


        print("\n--- Robot state ---")
        print("Position:", agent["position"])
        print("Rotation:", agent["rotation"])
        print("Camera horizon:", agent["cameraHorizon"])
    
        print("\n--- Manipulator state ---")
        print("Hand / end-effector position:", arm["handSphereCenter"])    
        print("Hand sphere radius:", arm["handSphereRadius"])
        
    # ==================================================
    # Show errors if an action fails
    # ==================================================

    event = controller.last_event

    if not event.metadata["lastActionSuccess"]:

        print(
            "Action failed:",
            event.metadata["errorMessage"]
        )

cv2.destroyAllWindows()
controller.stop()
```
